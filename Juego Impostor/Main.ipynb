```python

# 1. Modelo y Vista
%run Modelo.ipynb
%run Vista.ipynb

# El Controlador hace "from modelo import ...".
# En un cuaderno no existe el archivo modelo.py, así que le indicamos que "modelo"
# es este mismo cuaderno (donde ya se ejecutó el código del modelo).
import sys
sys.modules["modelo"] = sys.modules["__main__"]

# 2. Controlador
%run Controlador.ipynb

class VistaConsola:
    # El controlador espera una vista con estos métodos:
    #   limpiar_pantalla, esperar_enter, elegir_opcion, pedir_texto,
    #   mostrar_mensaje, mostrar_error, mostrar_pistas,
    #   mostrar_rol_privado y mostrar_resultado.
    # Esta clase los ofrece usando lo que ya existe en la Vista.
    
    def __init__(self):
        self._vista_ronda = VistaRondaConsola(esperar_debate=False)
        self._vista_partida = VistaPartidaConsola()

    def limpiar_pantalla(self):
        limpiar_pantalla()

    def esperar_enter(self, mensaje):
        input(mensaje)

    def elegir_opcion(self, pregunta, opciones):
        # Muestra la lista numerada y devuelve la POSICIÓN elegida (0, 1, 2...)
        print(f"\n{pregunta}")
        numero = 1
        for opcion in opciones:
            print(f"  {numero}. {opcion}")
            numero = numero + 1
        elegida = leer_entero("Tu opción: ", 1, len(opciones))
        return elegida - 1

    def pedir_texto(self, pregunta):
        return input(pregunta + " ").strip()

    def mostrar_mensaje(self, mensaje):
        print(mensaje)

    def mostrar_error(self, mensaje):
        self._vista_ronda.mostrar_error(mensaje)

    def mostrar_pistas(self, pistas):
        self._vista_ronda.mostrar_pistas(pistas)

    def mostrar_rol_privado(self, jugador):
        rol = jugador.get_rol()
        print(f"  Rol:     {rol.nombre}")
        print(f"  Palabra: {rol.obtener_palabra_visible()}")

    def mostrar_resultado(self, resultado):
        self._vista_partida.mostrar_resultado_final(resultado)


def main():
    controlador = ControladorPartida(
        VistaConsola(),
        RepositorioPalabrasMemoria(),
        AsignadorRolesAleatorio(),
        ContadorVotosMayoria(),
    )
    controlador.jugar()


main()
```
