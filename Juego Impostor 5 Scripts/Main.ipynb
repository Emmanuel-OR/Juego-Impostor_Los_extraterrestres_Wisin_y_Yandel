```python
%run Modelo.ipynb
%run Servicios.ipynb
%run Vista.ipynb

import sys
sys.modules["modelo"] = sys.modules["__main__"]

%run Controlador.ipynb


class VistaConsola:

    def __init__(self):
   
        self._vista_ronda = VistaRondaConsola(esperar_debate=True)
        self._vista_partida = VistaPartidaConsola()

    def limpiar_pantalla(self):
        limpiar_pantalla()

    def esperar_enter(self, mensaje):
        input(mensaje)

    def elegir_opcion(self, pregunta, opciones):
        print(f"\n{pregunta}")
        numero = 1
        for opcion in opciones:
            print(f"  {numero}. {opcion}")
            numero = numero + 1
        elegida = leer_entero("Tu opción: ", 1, len(opciones))
        return elegida - 1

    def pedir_entero(self, pregunta, minimo, maximo):
        return leer_entero(pregunta + " ", minimo, maximo)

    def pedir_si_no(self, pregunta):
        while True:
            respuesta = input(pregunta + " (s/n): ").strip().lower()
            if respuesta in ("s", "si", "sí", "SI", "SÍ", "Si", "S"):
                return True
            if respuesta in ("n", "no", "NO", "No", "N"):
                return False
            print("  Responde 's' o 'n'.")

    def pedir_texto(self, pregunta):
        return input(pregunta + " ").strip()

    def mostrar_mensaje(self, mensaje):
        print(mensaje)

    def mostrar_error(self, mensaje):
        self._vista_ronda.mostrar_error(mensaje)

    def mostrar_pistas(self, pistas):
        self._vista_ronda.mostrar_pistas(pistas)

    def esperar_debate(self, segundos):
        self._vista_ronda.esperar_debate(segundos)

    def mostrar_rol_privado(self, jugador, categoria):
        rol = jugador.get_rol()
        print(f"  Rol:       {rol.nombre}")
        print(f"  Categoría: {categoria}")
        print(f"  Palabra:   {rol.obtener_palabra_visible()}")

    def mostrar_resultado(self, resultado):
        self._vista_partida.mostrar_resultado_final(resultado)


def main():
    controlador = ControladorPartida(
        VistaConsola(),
        RepositorioPalabrasMemoria(),
        AsignadorRolesAleatorio(),
        ContadorVotosMayoria(),
    )
    controlador.jugar()


main()
```
